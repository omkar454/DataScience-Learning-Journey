```python
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

# Union-:
union_set = set1.union(set2)
union_set
```

```python
# Intersection-:
intersection_set = set1.intersection(set2)
print(intersection_set)
```

```python
A = {1, 2, 3}
B = {3, 4, 5}

print(A | B)   # Union
print(A & B)   # Intersection
print(A - B)   # Difference
print(A ^ B)   # Symmetric difference

# Python provides operators |, &, -, ^ as alternatives to set methods for union, intersection, difference, and symmetric difference.
```
