```python
# Shallow copy shares nested objects, while deep copy duplicates everything.
# deepcopy-:
# import copy
# d2 = copy.deepcopy(d1)
# Explanation:
# Creates a completely independent copy, including all nested objects.
```

```python
# Shallow Copy
student_copy = student.copy()
print(student_copy)
print(student)
```

```python
student["name"] = "Omkar Raut"
print(student_copy)
print(student)
```

```python
# Shallow and deep copy matter only for mutable types; without copying, mutable objects are fully shared, while immutable objects are unaffected.

# ✅ Mutable data types-:
# list,dict,set,bytearray,user-defined objects (class instances)

# ✅ Immutable data types-:
# int,float,complex,bool,str,tuple,frozenset,bytes
```

```python
for keys in student.keys():
    print(keys)
```

```python
for items in student.items():
    print(items)
```

```python
for values in student.values():
    print(values)
```

```python
for key,value in student.items():
    print(f"{key} : {value}")
```

```python
students = {
    "student1":{"name":"krish","age":32},
    "student2":{"name":"Peter","age":35}
}
print(students)
```

```python
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
for student_Id, student_Info in students.items():
    print(f"{student_Id}:{student_Info}")
    for key,value in student_Info.items():
        print(f"{key}:{value}")
```

```python
# Dictionary Comprehension-:
squares = {x:x**2 for x in range(5)}
print(squares)
```

```python
evens = {x:x**2 for x in range(10) if  x%2==0}
print(evens)
```

```python
# Practical Examples-:
numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number] += 1
    else:
        frequency[number] = 1

print(frequency)
```
