## List Comprehension

```python
lst = []
for x in range(10):
    lst.append(x**2)
print(lst)
```

```python
[ x**2 for x in range(10)]
```

Basics Syantax [expression for item in iterable]

with conditional logic [expression for item in iterable if condition]

Nested List Comprehension [expression for item1 in iterable1 for item2 in iterable2]

```python
square = [num**2 for num in range(10)]
```

```python
square
```

```python
even_num = [i for i in range(10) if i%2==0]
```

```python
even_num
```

```python
lst1=[1,2,3,4]
lst2=['a','b','c','d']

pair = [[i,j] for i in lst1 if i>2 for j in lst2]
print(pair)
```

A list comprehension with only if is used to filter elements from an iterable. No else is allowed in such list comprehensions

```python
## List Comprehension with function calls
words = ["hello", "world", "python", "list", "comprehension"]
lengths = [len(word) for word in words]
print(lengths)  # Output: [5, 5, 6, 4, 13]
```

When else is also used, the list comprehension becomes a conditional expression (not just filtering)

Only if → filters elements

if–else → transforms every element

### Structure to remember
[value_if_true if condition else value_if_false for item in iterable]

```python
nums = [1, 2, 3, 4, 5]

result = ["Even" if n % 2 == 0 else "Odd" for n in nums]
print(result)
```
