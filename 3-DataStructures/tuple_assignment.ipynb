### Assignment 15: Tuple Functions

Write functions that take a tuple and return the minimum, maximum, and sum of the elements. Print the results for a sample tuple.

```python
def tuple_op(tup):
    sum = 0
    min = tup[0]
    max = tup[0]
    for item in tup:
        sum += item
        if item > max:
            max= item
        if item < min:
            min = item
    print(f"Max:{max} ; Min:{min}; Sum:{sum}")

tuple_op((1,2,3,0,-2))
```
