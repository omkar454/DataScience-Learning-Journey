### Assignment 1: Reading a File

Write a function that reads the contents of a file named `sample.txt` and prints each line.

```python
def read_file(file_path):
    with open(file_path,"r") as file:
        for line in file.readlines():
            print(line.strip("\n"))

read_file("sample.txt")
```

### Assignment 3: Copying a File

Write a function that copies the contents of a file named `source.txt` to a new file named `destination.txt`.

```python
def src_dst(src,dst):
    with open(src,"r") as file:
        content = file.read()

    with open(dst,"w") as file:
        file.write(content)

src_dst("source.txt","destination.txt")
```

```python
def appender(file_path, content):
    with open(file_path,"a") as file:
        file.write('\n'+content+'\n')

appender("log.txt", "Appending this line to a new line")
```

### Assignment 5: Counting Words in a File

Write a function that reads the contents of a file named `document.txt` and returns the number of words in the file.

```python
def word_counter(file_path):
    total_words=0
    with open(file_path,"r") as file:
        lines = file.readlines()
        for line in lines:
            words = line.strip("\n.").split()
            print(words)
            total_words += len(words)
    return total_words

print(word_counter("sample.txt"))
            
```

### Assignment 6: Finding and Replacing Text

Write a function that finds and replaces all occurrences of a given word in a file named `data.txt` with another word.

```python
def find_reaplace(file_name,old_word, new_word):
    with open(file_name, "r") as file:
        content = file.read()
        with open(file_name,"w") as replace:
            print(content)
            final_content = content.replace(old_word,new_word)
            replace.write(final_content)

find_reaplace("data.txt", "hello","hi")

        
```

### Assignment 7: Reading a File in Reverse

Write a function that reads the contents of a file named `reverse.txt` and prints each line in reverse order.

```python
def reverser(file_path):
    with open(file_path,"r") as file:
        content = file.readlines()
        print(content)
        for line in content[::-1]:
            line = line.strip("\n").strip()
            print(line)

reverser("data.txt")
```

### Assignment 8: Counting Lines, Words, and Characters

Write a function that reads the contents of a file named `stats.txt` and returns the number of lines, words, and characters in the file.

```python
def line_word_char(file_path):
    total_char = 0; total_words = 0
    with open(file_path,"r") as file:
        lines = file.readlines()
        total_lines = len(lines)
        for line in lines:
            words = line.strip("\n").split()
            total_words += len(words)
            for word in words:
                    total_char += len(word)
    return total_lines, total_words, total_char

line_word_char("data.txt")
```

### Assignment 9: Merging Multiple Files

Write a function that merges the contents of multiple files into a single file named `merged.txt`.

```python
def merged_file(file1,file2, merged_file):
    with open(file1,"r") as file_1:
        content1 = file_1.read()
        with open(file2,"r") as file_2:
            content2 = file_2.read()
            with open(merged_file,"w") as merger:
                merger.write(content1 + "\n" + content2)

merged_file("log.txt","output.txt","merged.txt")
```

### Assignment 10: Splitting a Large File

Write a function that splits a large file named `large.txt` into smaller files of 100 lines each.

```python
import math
def splitter(large_file):
    temp_lines = 0
    with open(large_file,"r") as large:
        lines = large.readlines()
        total_files = math.ceil(len(lines)/100)
        print(total_files)
        for i in range(total_files):
            if i == total_files -1:
                 end_line = len(lines) - temp_lines
            else:
                 end_line = 100
            for j in range(end_line):
                with open(f"small{i}.txt","a") as small:
                        small.write(lines[temp_lines])
                        temp_lines += 1

splitter("large.txt")
    

        
```

### Assignment 11: Creating a Log File

Write a function that creates a log file named `activity.log` and writes log messages with timestamps.

```python
from datetime import datetime
import time
def logginge_message(filepath,log_msg):
    with open(filepath,"w") as logger:
        for info,msg in log_msg.items():
            logger.write(f"{datetime.now().isoformat()} - {info} - {msg}\n")
            time.sleep(2)
            
log_msg = {
    "INFO":"Program executing",
    "WARN":"Error in Program",
    "SUCCESS":"Program execution completed"
}
logginge_message("activity.log",log_msg)

```

### Assignment 12: Binary File Operations

Write a function that reads a binary file named `image.bin` and writes its contents to another binary file named `copy_image.bin`.

```python
def copy_paste_bin(file1,file2):
    with open(file1,"wb+") as file:
        file.write(b"\x41\xFF\x10\x3A\x00\x7C\xA9\x5F\xC3\x18\x9D\x02\xE7\x44\xB1\x6D")
        file.seek(0)
        content  = file.read()
        print(content)
        with open(file2,"wb") as copy_file:
            copy_file.write(content)

copy_paste_bin("image.bin","copy_image.bin")        
```

### Assignment 13: CSV File Operations

Write a function that reads a CSV file named `data.csv` and prints its contents as a list of dictionaries.

```python
import csv

def csv_paster(file_path):
    with open(file_path,"w+", newline="") as file:
        writer = csv.DictWriter(file,fieldnames=["Age","Name","Marks"])
        writer.writeheader()
        writer.writerows([
            {"Age":12,"Name":"Omkar","Marks":20},
            {"Age":22,"Name":"Suresh","Marks":22},
            {"Age":32,"Name":"Raut","Marks":40}
                          ])
        file.seek(0)
        reader = csv.DictReader(file)
        print(list(reader))
        
csv_paster("data.csv")
```

### Assignment 14: JSON File Operations

Write a function that reads a JSON file named `data.json` and prints its contents as a Python dictionary.

```python
import json

def json_paster(file_path):
    with open(file_path,"w+") as file:
        data = [{
            'name':'Omkar',
            'age':20
        },
        {
            'name':'Suresh',
            'age':17
        }]
        file.write(json.dumps(data))
        file.seek(0)
        print(json.loads(file.read()))

json_paster("data.json")

```

### Assignment 15: File Permission Handling

Write a function that attempts to read a file named `protected.txt` and handles any permission errors gracefully by printing an error message.

```python
def read_file(filePath):
    try:
        with open(filePath,"r") as file:
            print(file.read())
    except PermissionError as e:
        print(f"You do not have permission to open this file: {e}")

read_file("protected.txt")
```
