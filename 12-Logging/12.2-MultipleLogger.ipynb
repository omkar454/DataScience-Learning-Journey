```python
import logging
```

```python
# Create a logger for module1-:
logger1 = logging.getLogger("module1")
# setLevel() is used to set the threshold level for a logger or handler.
# Only log messages at that level or higher will be processed.
logger1.setLevel(logging.DEBUG)

# Create a logger for module2-:
logger2 = logging.getLogger("module2")
logger2.setLevel(logging.WARNING)
```

```python
logging.basicConfig(
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%M-%d %H:%M:%S'
)
```

```python
logger1.debug("This is debug message for module1")
logger2.warning("This is warning message for module2")
logger2.error("This is error message for module2")
```
