````python
# SMOTE(Syntheic Minority Oversampling Technique) creates new fake data points for the minority class instead of copying old ones.[TYPE OF UPSAMPLING TECHNIQUE]

# Simple idea:
# Dataset is imbalanced (few minority samples)
# SMOTE looks at nearby minority points
# Then creates new points between them

# So instead of duplicating data:
# ❌ Copy-paste rows
# ✔ Make new realistic synthetic rows

# One-line version:
# SMOTE balances data by creating new artificial minority samples using existing ones.

# Simple example:
# If minority points are:
# (2,3) and (4,5)
# SMOTE creates a new point like:
# (3,4)

# In short:
# SMOTE makes new smart samples, not duplicates, to balance the dataset.
````

````python
# Synthetic means:
# Artificially created (not real-world data).

# In simple words:
# Data that is made by a program, not collected from real life.

# Unsynthentic / Non-synthetic means:
# Real, natural, original data collected from real sources.

# In simple words:
# Data that comes from real-world measurements or records, not artificially generated.
````

````python
# make_classification
# Generates a synthetic classification dataset for machine learning.

# In short:
# It creates fake labeled data for training and testing ML models
````

````python
# ### How can ONE class have MULTIPLE clusters?

# **Class ≠ Cluster**

# They are different things:

# * **Class** → label (0 or 1, cat/dog, spam/not spam)
# * **Cluster** → group of similar points in space

# ## Simple intuition:

# A single class can be **spread in different regions** of the feature space.

# ### Example:

# Class = **“Students who passed”**

# But they can form clusters like:

# * Group 1: High math, high science
# * Group 2: High English, high history

# Same class ✅
# Different clusters ✅

# ---

# ## In `make_classification`:

# ```python
# n_clusters_per_class = 3
# ```

# Means:
# 👉 Each class will be split into **3 separate groups** in feature space.

# So for 2 classes:

# * Class 0 → 3 clusters
# * Class 1 → 3 clusters

# Total = **6 clusters**, but still **2 labels**


# ## Visual example idea:

# ```
# Class 0:   ●●●       ●●●       ●●●
# Class 1:      ▲▲▲       ▲▲▲       ▲▲▲
# ```

# Same class symbols, different locations.

# ### In short:

# > A class can have many clusters because clusters represent **data distribution**, not labels.
````

````python
# Feature Terms in make_classification:

# n_features → Total number of features (columns)
# Condition: n_features > 0
# Default: 20

# n_informative → Useful features for prediction
# Condition: n_informative ≤ n_features
# Default: 2

# n_redundant → Derived/duplicate-type features
# Condition:
# n_informative + n_redundant + n_repeated ≤ n_features
# Default: 2

# n_repeated → Exact copy features
# Condition:
# n_informative + n_redundant + n_repeated ≤ n_features
# Default: 0

# Noise features → Random/useless features (auto-generated)
# There is NO parameter in make_classification to generate NOISE FEATURES. 
# Condition:
# n_features > (n_informative + n_redundant + n_repeated)
# Default: Auto-created

# Eligibility Rule:
# n_informative + n_redundant + n_repeated ≤ n_features
# Noise Rule:
# n_features > (n_informative + n_redundant + n_repeated)

# One-line summary:

# All feature types are subsets of n_features, and extra features become noise automatically.
````

````python
from sklearn.datasets import make_classification
# X -> Independent Features (I/P Features)
# y -> Dependent Feature(Singular) (O/P Feature)
X,y = make_classification(
    n_samples=1000,
    n_features=2,
    n_redundant=0,
    n_clusters_per_class=1,
    weights=[0.90],
    random_state=12,
)

# n_samples=1000
# 👉 Create 1000 data points (rows)
# n_redundant=0
# 👉 No duplicate/repeated features
# (no feature is just a copy/combination of another feature)
# n_features=2
# 👉 Each data point has 2 columns (features)
# n_clusters_per_class=1
# 👉 Each class forms 1 cluster only
# weights=[0.90]
# 👉 Imbalanced data

# 90% → Class 0
# 10% → Class 1 (minority class)

# random_state=12
# Fixes randomness (same dataset every time)
````

````python
# **`n_redundant` means:**
# 👉 Number of **duplicate / derived features** in the dataset.

# ### Simple words:
# > Features that **don’t add new information** because they are made from other features.

# ### Example:
# If you have:
# * `height_cm`
# * `height_m`

# One is just a converted version of the other → **redundant**
# ### In `make_classification`:

# ```python
# n_redundant=0
# ```

# Means:
# 👉 **No redundant features**
# All features are **independent and useful**

# **In short:**
# > `n_redundant` controls how many **repetitive (non-useful) features** are created.
````

SMOTE Implementation-:

````python
# ```python
# oversample = SMOTE()
# ```

# ➡️ Creates a **SMOTE object**
# SMOTE = *Synthetic Minority Over-sampling Technique*
# It’s used to fix **class imbalance** by generating fake (synthetic) data points for the minority class.

# ```python
# X, y = oversample.fit_resample(final_df[['f1','f2']], final_df['target'])


# ### What happens here:
# 1. `final_df[['f1','f2']]` → feature matrix **X**
# 2. `final_df['target']` → label vector **y**
# 3. `fit_resample(...)` does two things:

#    * **fit** → learns minority class structure (nearest neighbors)
#    * **resample** → creates **synthetic samples** of minority class
# 4. Returns:

#    * `X` → new balanced features
#    * `y` → new balanced labels


# ### Simple meaning:

# > This code **balances the dataset** by generating new artificial minority-class points using SMOTE so that both classes have equal representation.


# ### In ultra-short form:
# > Converts imbalanced data → balanced data using synthetic samples.
````

````python
from imblearn.over_sampling import SMOTE
oversample = SMOTE()
X,y = oversample.fit_resample(final_df[['f1','f2']],final_df['target'])
````
