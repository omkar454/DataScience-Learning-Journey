```python
# Target Guided Ordinal Encoding is a manual ordinal technique.
```

```python
import pandas as pd
df = pd.DataFrame({
    'city':['New York','London','Paris','Tokyo','New York','Paris'],
    'price':[200,150,300,250,100,320]
})
```

```python
df
```

```python
mean_map = df.groupby('city')['price'].mean().to_dict()
```

```python
mean_map
```

```python
df['city_encoded'] = df['city'].map(mean_map)
```

```python
df
```

```python
df[["price","city_encoded"]]
```

```python
# Here’s a clear explanation:

# Monotonic:

# A relationship that always moves in one direction — either only increasing or only decreasing.

# Example: Age vs. Income (if income consistently rises with age) → monotonic increasing.

# Non-monotonic:

# A relationship that changes direction — sometimes increases, sometimes decreases.

# Example: Hours of study vs. Stress (stress may increase at first, then decrease after a point) → non-monotonic.

# In short:

# Monotonic = always one-directional, Non-monotonic = direction changes.
```

```python
import seaborn as sns
df_tip = sns.load_dataset('tips')
```

```python
df_tip
```

```python
# observed in groupby() means:
# It controls whether pandas groups only existing categories or all possible categories of a categorical column.

# Simple:
# observed=False → include all categories (even those not present in data)
# observed=True → include only categories that actually appear

# In one line:
# observed decides if grouping uses all categories or only observed (present) categories.
```

```python
time_map = df_tip.groupby("time", observed=True)["total_bill"].mean().round(2).to_dict()
```

```python
df_tip["time_encoded"] = df_tip["time"].map(time_map)
```
