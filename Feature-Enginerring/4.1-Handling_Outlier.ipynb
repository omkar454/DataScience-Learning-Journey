```python
# Minimum, Maximum, Median, Q1, Q3, IQR
```

```python
import numpy as np
import seaborn as sns
```

```python
lst_marks = [42,32,56,75,89,90,87,67,54,45,98,99,67,74]
# It computes the minimum, 25th percentile (Q1), median (Q2), 75th percentile (Q3), and maximum values of lst_marks.
# (Conceptually: quantile summary of the data distribution)
minimum,Q1,median,Q3,maximum = np.quantile(lst_marks,[0,0.25,0.50,0.75,1])
```

```python
minimum,Q1,median,Q3,maximum
```

```python
IQR = Q3 - Q1
print(IQR)
```

```python
lower_fence = Q1 - (1.5*IQR)
higher_fence = Q3 + (1.5*IQR)
```

```python
lower_fence
```

```python
higher_fence
```

```python
# Any data less than the Lower Fence (LF) or greater than the Upper Fence (HF) is an outlier.
# Data equal to LF or HF is not considered an outlier.
```

```python
sns.boxplot(lst_marks)
```
