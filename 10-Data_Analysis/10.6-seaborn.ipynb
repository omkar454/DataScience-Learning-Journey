```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Basic Plottings With Seaborn-:
sns.get_dataset_names()
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
# Create a scatter plot-:
sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title("Scatter plot of Total Bill vs Tip")
plt.show()
```

```python
# Line Plot
sns.lineplot(x='size',y='total_bill',data=tips)
plt.title("Line Plot of Total bill by size")
plt.show()
```

```python
# The dark blue line = mean (average) total_bill for each group size
# The light blue shaded area = 95% confidence interval
#   → It shows the uncertainty around the mean
#   → The true average usually lies within this shaded region
#   → Dark line = best estimate, shading = range where true mean likely falls
```

```python
# Categorical Plots
# Bar Plot
sns.barplot(x='day',y='total_bill',data=tips)
plt.title("Bar Plot of Total Bill By day")
plt.show()
```

```python
# NOTE:
# - The top of each blue bar represents the SAMPLE MEAN of total_bill for that day.
# - The vertical black line on each bar is the ERROR BAR.
# - Seaborn automatically adds a 95% CONFIDENCE INTERVAL (CI) around the mean.
# - This CI shows the RANGE in which the TRUE POPULATION MEAN is expected to lie with 95% confidence.
# - If the black line is tall → more variation in data.
# - If the black line is short → the data is more stable.
# - Blue bar = best estimate of mean.
# - Black line = uncertainty around that mean.
```

```python
# Box Plot-:
sns.boxplot(x='day',y='total_bill',data=tips)
```

```python
# Box = middle 50% data, line inside = median, whiskers = range, dots = outliers.
```

```python
# Violin Plot-:
sns.violinplot(x='day',y='total_bill',data=tips)
```

```python
# NOTE:
# - A violin plot shows both the distribution shape AND summary statistics.
# - The OUTER BLUE CURVES represent the KDE (Kernel Density Estimation):
#       → Wider sections = more data points (dense area)
#       → Narrow sections = fewer data points
#       → Smooth shape = like a smoothed histogram
#
# - The WHITE DOT inside the violin is the MEDIAN (50th percentile).
# - The THICK BLACK BAR shows the IQR (25th to 75th percentile).
# - The THIN BLACK LINE shows the overall data range (min to max without outliers).
#
# - Violin plot = Box plot + Smooth density curve (distribution) combined.
# - Helps to see where values concentrate and how data is spread for each day.
```

```python
# Histogram-:
sns.histplot(tips['total_bill'], bins=10,kde=True)
```

```python
# KDE Plot-:
sns.kdeplot(tips['total_bill'],fill=True)
```

```python
# Pair Plot-:
sns.pairplot(tips)
```

```python
sns.pairplot(tips,hue="day")
```

```python
import pandas as pd
sales_df = pd.read_csv("sales_data.csv")
sales_df.head()
```

```python
# Plot total sales  average  by product
plt.figure(figsize=(10,6))
sns.barplot(x='Product Category',y='Total Revenue', data=sales_df,estimator="mean")
```

```python
# Plot total saled  by product
plt.figure(figsize=(10,6))
sns.barplot(x='Product Category',y='Total Revenue', data=sales_df,estimator=sum)
plt.title("Total sales by Product")
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.show()
```

```python
# CI exists because your data is a sample, not the whole population.
# It shows uncertainty, not mistakes.
# Bigger data → smaller CI → more confidence

# CI is based on the sample.
# CI is about the population.
# Both are true

# Confidence Interval (CI) gives a range where the true value for the whole world (population) is likely to be.
# It is calculated from your sample, but it represents the entire population.
# A 95% CI means: “If we took many samples, 95% of them would contain the true population value.”

# 95% confidence means:
# “We are 95% confident that the true population value lies inside this CI interval.”
# It does not mean 95% probability.
# It means the method captures the true value 95% of the time when repeated many times.

# But for practical understanding:
# ✅ You can say:
# “We have 95% confidence that the true value is inside this CI interval.”
```

```python
## Plot total sales by Region
plt.figure(figsize=(4,4))
sns.barplot(x='Region',y="Total Revenue",data=sales_df,estimator=sum)
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.show()
```

```python
# In plt.figure(figsize=(4,2)):
# 4 = width of the figure
# 2 = height of the figure
# Both are measured in inches.
# So your graph will be 4 inches wide and 2 inches tall.
```
