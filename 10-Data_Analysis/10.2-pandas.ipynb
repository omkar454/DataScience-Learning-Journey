```python
import pandas as pd
```

```python
# Series-:
data = [1,2,3,4,5]
series = pd.Series(data)
print(series)
print(series[0])
print(type(series))
```

```python
# Series from dictionary-:
data = {'a':1,'b':2,'c':3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
data=[10,20,30]
index=['a','b','c']

pd.Series(data, index=index)
```

```python
# DataFrame-:
# Dataframe from dictionary of list-:

data = {
    'Name':['Omkar','John','Jack'],
    'Age':[25,30,45],
    'City':['Mumbai','Kota','Jaipur']
}

df = pd.DataFrame(data)
print(df)
print(df['Name'][0])
print(df['Name'][1])
print(df['Age'][0])
print(df['Age'][1])
print(type(df))
```

```python
import numpy as np
np.array(df)
```

```python
# DataFrame from a List of Dictionaries-:
data = [
 {'Name':'Omkar','Age':25,'City':'Mumbai'},
 {'Name':'Raut','Age':35,'City':'Jaipur'},
 {'Name':'Suresh','Age':22,'City':'Firang'}
]

df = pd.DataFrame(data)
print(df)
print(type(df))
```
