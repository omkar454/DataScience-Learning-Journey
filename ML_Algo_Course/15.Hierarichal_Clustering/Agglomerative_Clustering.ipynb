```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
```

```python
iris = datasets.load_iris()
```

```python
iris
```

```python
iris_data = pd.DataFrame(data=iris['data'],columns=iris['feature_names'])
```

```python
iris_data
```

Since; here we are doing CLUSTERING and NOT PREDICTION of any test-data.
So; we will use ONLY I/P features and ignore the TARGET/Ouput Feature.
As; we need to find hidden patterns from I/P features and we do NOT need O/P feature for finding those hidden patterns according to this situation. So; we will NOT use extra feature-: O/P feature which is REDUNDANT feature in this case.






We want the algo to find hidden patterns and not to group based on O/P feature as this grouping is already known to us. SO; if we add O/P feature then it will NOT result in finding hidden patterns but will find  a known PATTERN instead.




# So; never ever include O/P feature [TARGET FEATURE] for CLUSTERING / UNSUPERVISED M.L. Algorithm as for clustering -: O/P feature is a redundant feature. So; never use O/P feature for CLUSTERING But in very rare situations this O/P feature can be used for VALIDATION of PERFORMANCE of the CLUSTERING ALGO's..

```python
# We need to do STANDARDIZATION as Agglomerative and Divisive clusterings are distance based unsupervised M.L. algorithms
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(iris_data)
```

```python
X_scaled
```

```python
# Now we will apply PCA(Principle Component Analysis) unsupervised dimensional reduction M.L. technique to reduce the number of dimenaions/features in order to visualize the clustering done by Agglomerative_Clustering in 2D as total number of I/P features currently in IRIS datset = 4 so we cannot visulaize and also even cannot PLOT 4D graph

# So; using PCA we will reduce features/dimensions to "2" So; we can plot and even visualize the  clustering done by Agglomerative_Clustering M.L. algorithm.

from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
```

```python
pca
```

```python
pca_scaled = pca.fit_transform(X_scaled)
```

```python
# Reducing 4D I/P to 2-Dimensions/Features I/P
pca_scaled
```

```python
plt.scatter(pca_scaled[:,0],pca_scaled[:,1],c=iris.target)
```

So; by scatter plot we can clearly see there are 3 proper clusters which are formed for the datapoints of the dataset with respect to the target/OutPut feature.

So; by using Agglomerative Clustering -: we will see whether we will get k =3 or NOT.
So; indirectly we are using O/P feature as validation for performance of clustering algo. but will NEVER USE the O/P feature for training of clustering algorithms as this would lead to data/answer leakage.






























Leaf nodes in a dendrogram = Original data points in the dataset

So; Number of leaf nodes in a dendrogram = Number of original data points in the dataset

These are majorly and widely used parameters of scipy.cluster.hierarchy.linkage() method

```python
# Agglomerative CLustering
# Step 1: Construct a DENDROGRAM
import scipy.cluster.hierarchy as sc
# plot the dendrogram-:
plt.figure(figsize=(20,7))
plt.title("Dendrograms")

# create dendrogram by using linkage matrix returned by scipy.cluster.hierarchy.linkage() method and scipy.cluster.hierarchy.dendrogram() for creation of Dendrogram via this linkage matrix-:
sc.dendrogram(sc.linkage(pca_scaled,method="ward",optimal_ordering=True))
plt.xlabel("Sample Index/ Data-point numbers/ row numbers of the dataset")
plt.ylabel("Euclidean Distance")
plt.show()
```

Since ; looking at DENDROGRAM the longest vertical thorugh which NO horizontal line passes through it is the RIGHT VERTICAL LINE of the biggest building of dendrogram and if we pass a horizontal line through it then that HORIZONTAL LINE will pass thorugh 2 VERTICAL LINES. So; most optimal k value = 2 for Agglomerative Clustering on this dataset according to this above DENDROGRAM.

```python
# Linkage matrix returned by scipy.cluster.hierarchy.linkage() method as shown below which was used above for creation of DENDROGRAM-:
sc.linkage(pca_scaled,method="ward",optimal_ordering=True)
# 1.01000000e+02 means 1.01000000 × 10², which equals 101.
# 1.01000000e-02 means 1.01000000 × 10⁻², which equals 0.0101.
```

```python
sc.linkage(pca_scaled,method="ward",optimal_ordering=True).shape
# total datapoints = 150
# Total merges of clusters required to go from 150 individual clusters 
# [1 datapoint => 1 cluster] to single huge cluster consisting of all 150 data-points =149 merges of clusters
# so due to this reason we got O/P = (149, 4)
```

# From the DENDROGRAM -: most optimal k value we got is "2" So ; we will keep n_clusters=2 for Agglomerative Clustering

Widely used parameters of Agglomerative Clustering-:






If distance_threshold is used, n_clusters must be None [n_clusters = None] because the number of clusters is automatically determined by the distance threshold.



```python
from sklearn.cluster import AgglomerativeClustering
cluster = AgglomerativeClustering(n_clusters=2,metric="euclidean",linkage='ward')
cluster.fit(pca_scaled)
```

Widely used attributes of Agglomerative Clustering-:








n_leaves_ : number of original data points in dendrogram

Leaves = end nodes of dendrogram representing each data point before merging == Points on X-axis of Dendrogram



```python
cluster.n_clusters_
```

```python
cluster.labels_
```

```python
plt.scatter(pca_scaled[:,0],pca_scaled[:,1],c=cluster.labels_)
```

So; accroding to original dataset the total number of clusters should be "3" for  IRIS DATASET but Agglomerative Clustering according to its DENDROGRAM creation it suggested that NO NEED of 3 clusters instead keep n_clusters="2" and still you will get PROPER CLEAR SEPERABLE 2 clusters on the GRAPH as shown above.
