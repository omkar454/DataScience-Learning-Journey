```python
from sklearn.datasets import load_iris
```

```python
from sklearn.model_selection import train_test_split
```

](attachment:image.png)

```python
X,y = load_iris(return_X_y=True)
```

```python
# return_X_y=True → The function returns only the data and target arrays:
# X → feature values (shape [n_samples, n_features])
# y → target labels (shape [n_samples])
# No extra info like feature names, target names, or a description is returned.
```

```python
X
```

```python
y
```

```python
load_iris()
```

```python
load_iris().keys()
```

```python
X_train, X_test, y_train, y_test  = train_test_split(X,y,test_size=0.3,random_state=0)
```

```python
X_train
```

Since; all the I/P features are continuos variables.
So; we will use Gaussian Naive-Bayes-Distribution on this dataset.
Also each feature of IRIS DATASET is always NORMALLY/GAUSSIAN DISTRIBUTED

So; use Gaussian naive bayes distribution here

```python
from sklearn.naive_bayes import GaussianNB
```

```python
gnb = GaussianNB()
```

```python
gnb.fit(X_train,y_train)
```

```python
y_pred = gnb.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
print(confusion_matrix(y_pred,y_test))
```

```python
accuracy_score(y_pred,y_test)
```

```python
print(classification_report(y_pred,y_test))
```

We got 100% accuracy cause iris dataset is very very SMALL.

```python
from sklearn.preprocessing import LabelEncoder
le1_sex = LabelEncoder()
le2_smoker = LabelEncoder()
le3_time = LabelEncoder()
```

```python
df ['sex'] = le1_sex.fit_transform(df['sex'])
```

```python
df['smoker']=le2_smoker.fit_transform(df['smoker'])
```

```python
df['day'].unique()
```

```python
df['time']=le3_time.fit_transform(df['time'])
```

```python
from sklearn.preprocessing import OneHotEncoder
OHE = OneHotEncoder(drop='first')
```

```python
df['day'].unique()
```

```python
encoded = OHE.fit_transform(df[['day']])
```

```python
encoded
```

```python
print(encoded)
```



To convert sparse to dense use toarray() method-:



```python
encoded = encoded.toarray()
```

```python
OHE.get_feature_names_out()
```

```python
encoded
```

```python
import pandas as pd
ohe_df = pd.DataFrame(data=encoded,columns=OHE.get_feature_names_out())
```

```python
ohe_df
```

```python
df_final = pd.concat([ohe_df,df],axis=1)
```

```python
df_final.drop('day',inplace=True,axis=1)
```

```python
df_final.head()
```

```python
# Dependent features-:
X = df_final[['day_Sat','day_Sun','day_Thur','total_bill','tip','sex','smoker','size']]
# Independent features-:
y = df_final['time']
```

```python
X
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
from sklearn.naive_bayes import GaussianNB
# Gaussian Naive Bayes variant algo implementation
```

```python
X_train
```

```python
gnb = GaussianNB()
```

```python
gnb.fit(X_train,y_train)
```

```python
y_pred = gnb.predict(X_test)
# 0 -> Dinner
# 1 -> Lunch
```

```python
y_pred
```

```python
y_test
```

```python
print(confusion_matrix(y_pred,y_test))
print(accuracy_score(y_pred,y_test))
print(classification_report(y_pred,y_test))
```
