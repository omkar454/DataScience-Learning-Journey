


hstack -: horizontal stacking 




```python
# this is creating an inner-circle with eqn.: x**2 +  y**2 = 5**2 = 25
# center of circle is at origin= (0,0)
# rest all creation process is same as above outer circle
x1 = np.linspace(-5.0, 5.0, 100)
y1 = np.sqrt(5**2 - x1**2)
y1=np.hstack([y1,-y1])
x1=np.hstack([x1,-x1])
```

```python
# np.vstack([y,x])
# np.vstack([y,x])[0] ==> y
# np.vstack([y,x])[1] ==> x
# np.vstack([y,x]).shape ==> (2,200)
# np.vstack([y,x]).T ==> Transpose of np.vstack([y,x])
# np.vstack([y,x])[0] ==> x
# np.vstack([y,x])[1] ==> y
# np.vstack([y,x]).T.shape ==> (200,2)
```

```python
# hstack -: does stacking in HORIZONTAL way one after another i.e. after one list anohter list will be appended in the SAME list at end. NO creation of new LIST after one list.
# vstack -: does stacking in VERTICAL way one after another i.e. one list after another list. creation of new list after one list ends.
np.vstack([y,x]).T
# T -: transpose of multi-dimensional list/matrix -: np.vstack([y,x])
# Each row of points is a coordinate pair: (y, x)
# So y comes first, then x in each point
```

```python
df1 = pd.DataFrame(np.vstack([y,x]).T,columns=['X1','X2'])
df1['Y']=0
df2 = pd.DataFrame(np.vstack([y1,x1]).T,columns=['X1','X2'])
df2['Y']=1
```

```python
df1.head()
```

```python
df = pd.concat([df1,df2],ignore_index=True)
df
```

```python
df.tail()
```

```python
# Independent and dependent features-:
X = df.loc[:,:'X2']
y = df.loc[:,'Y']
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train ,y_test= train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
y_train
```

```python
# From this polynomial kernel-: the 3 important componenets which we will get from this kernel are-:
# 1) X1**2  2)X2**2  3)X1*X2
df['X1_squared'] = df['X1']**2
df['X2_squared'] = df['X2']**2
df['X1*X2'] = df['X2']*df['X1']
df.head()
```

```python
### Independent and Dependent features
X = df[['X1','X2','X1_squared','X2_squared','X1*X2']]
y = df['Y']
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=0,test_size=0.25)
```

```python
X_train
```





Plotly is a Python visualization library for creating interactive, web-based, dynamic graphs and dashboards.

```python
import plotly.express as px

fig = px.scatter_3d(df, x='X1', y='X2', z='X1*X2',
              color='Y')
fig.show()
```

Here; LINEAR SVC line is NOT possible cause points of both categories are NOT linearly seperated and few points of a class is on one side and other points are on toher side and in middle all points of one class is there.
So; linear SVC cannot be used.

So; wiht previous componenets-: X1; X2 and X1*X2. we cannot apply linear SVC for clear seperation of both the classes.

But; Now we will apply polynomial kernal which gave use 3 important components-: X1**2; X2**2 and X1*X2.
So; now we will plot the results of polynomial kernal via plotly-:

```python
import plotly.express as px

fig = px.scatter_3d(df, x='X1_squared', y='X2_squared', z='X1*X2',
              color='Y')
fig.show()
```

As u can see after applying POLYNOMIAL-KERNEL 
We get the plotting in a very clear and LINEAR SEPERABLE WAY cause of 3 components -: X1^2; X2^2 and X1*X2 provided by Polynomial kernel.

Now we can apply Linear SVC and get a Hyperplane(LINEAR HYPERPLANE) which will easily speprate both classes LINEARLY due to POLYNOMIAL KERNEL techique which will give us better accuracy -:


```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
classifier = SVC(kernel="linear")
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_test)
accuracy_score(y_test, y_pred)
```

We got 100% accuracy due to polynomial kernel.
But; here we manually calculated 3 componennts provided by Polynomial kernel and then added them in our dataset for training and testing and then we applied LINEAR KERNEL SVC and we got 100% accuracy.

But; instead of doing this manually if we used the dataset without manually calculating the components of polynomial kernel and just changing name of kernel to 'poly' then also we would have got the SAME accuracy as this entire process of calcualting components of polynomial kernel  would be done by polynomial kernel IMPLICITLY  like this-:

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
classifier = SVC(kernel="poly")
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_test)
accuracy_score(y_test, y_pred)
```

Same will work for rbf as well as sigmoid kernel also


They will do their respective transformations and their equations calculations internally only like this just by changing name of kernel to 'rbf' and 'sigmoid' respectively-:

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
classifier = SVC(kernel="rbf")
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_test)
accuracy_score(y_test, y_pred)
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
classifier = SVC(kernel="sigmoid")
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_test)
accuracy_score(y_test, y_pred)
```

This true power of SVM kernels.

Whenever u will get LOWER accuracy by Linear SVC kernel then change the kernal name to 'poly' or 'sigmoid' or 'rbf' -: u will definetly get better accuracy due to their transformation techniques which gives higher dimensional space and results in HIGHER ACCURACY.
