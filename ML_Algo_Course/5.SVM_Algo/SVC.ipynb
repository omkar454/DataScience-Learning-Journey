```python
# Now we will create a B.F.L with its set of marginal planes via LINEAR SVC as it is Binary Classification problem-:
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25,random_state=10)
```




cause high-dimensional data covers deatiled features of its DATASET.

```python
# Since; our dataset is CLEARLY/PROPERLY can be Linearly seperable
# So; we can use "linear" kernel for SVC-:
svc = SVC(kernel='linear')
```

```python
svc.fit(X_train,y_train)
```

```python
# Prediction -: 
y_pred = svc.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report(y_test,y_pred))
print('--------')
print(confusion_matrix(y_pred,y_test))
```

## Hyperparametere Tuning With SVC

```python
from sklearn.model_selection import GridSearchCV
 
# defining parameter range
param_grid = {'C': [0.1, 1, 10, 100, 1000],
              'gamma': [1, 0.1, 0.01, 0.001, 0.0001],
              'kernel': ['rbf']}
# kerel is only rbf here as it was GIVING the best accuracy as compared to LINEAR; polynomial and Sigmoid Kernels
```



Good for auditing

```python
grid = GridSearchCV(estimator=SVC(),param_grid=param_grid,cv=5,verbose=3,refit=True)
```

```python
grid
```

```python
X_train.shape, len(y_train)
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
# Prediction-:
y_pred = grid.predict(X_test)
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

After Hyper-Parameter Tuning with SVC 
We got a very good and HIGHEST accuracy  of 98% here
