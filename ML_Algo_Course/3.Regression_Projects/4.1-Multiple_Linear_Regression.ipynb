```python
# Independent and Dependent features-:
X = dataset.loc[:,:"Longitude"]
y = dataset.loc[:,"Price"]
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.33,random_state=10)
```

```python
X_train.shape,y_train.shape,X_test.shape,y_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
X_test
```

# Model Training

```python
from sklearn.linear_model import LinearRegression
```

```python
regression = LinearRegression()
```

```python
regression
```

```python
regression.fit(X_train,y_train)
```

```python
# slope or coefficients-: one coefficient for each feature.
# There are features -: So; total coefficients = 8
regression.coef_
```

```python
# intercept-:
regression.intercept_
```

```python
# Prediction for the tes data-:
y_pred = regression.predict(X_test)
```

```python
y_test
```

```python
y_pred
```

```python
# Performance Metrics-:
from sklearn.metrics import mean_squared_error,mean_absolute_error
print(mean_squared_error(y_test,y_pred))
print(mean_absolute_error(y_test,y_pred))
print(np.sqrt(mean_squared_error(y_test,y_pred)))
```



For numbers between 0 and 1, square root makes the value bigger

```python
# R square and adjusted R square-:
from sklearn.metrics import r2_score
score = r2_score(y_test,y_pred)
print(score)
```

```python
## Adjusted R square
#display adjusted R-squared
1 - (1-score)*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1)
```

Since; our dataset is NOT LINEARLY-DISTRIBUTED/SCATTERED .i.e there is NON-LINEAR-RELATIONSHIP -: So; we will get poor Performance Metrics

## Pickling 
Python pickle module is used for serialising and de-serialising a Python object structure. Any object in Python can be pickled so that it can be saved on disk. What pickle does is that it “serialises” the object first before writing it to file. Pickling is a way to convert a python object (list, dict, etc.) into a character stream. The idea is that this character stream contains all the information necessary to reconstruct the object in another python script.

```python
# 🔹 Pickling (in Python)
# Pickling means saving a Python object into a file so we can use it later.

# Example:
# Save trained ML model
# Save data or variables
# Later, we can load it back without training again.

# 🔹 Simple Meaning:
# Pickling = storing a Python object into a file for future use.
```

```python
import pickle
pickle.dump(regression,open("regressor.pkl","wb"))
```

```python
model = pickle.load(open("regressor.pkl","rb"))
```

```python
model.predict(X_test)
```
