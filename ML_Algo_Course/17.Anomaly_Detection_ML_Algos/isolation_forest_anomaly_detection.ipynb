```python
import pandas as pd
```

```python
df = pd.read_csv('healthcare.csv')
df.head()
```

This dataset consists of 2 features and these 2 features are represenation of health parameters of a human respectively and are talking about a certain disease.

So; not every person will have disease-: Only those records which are OUTLIERS will have disease and remaining records will be NOT having disease will be NORMAL POINTS in the dataset.

So; clearly we need to use ISOLATION-FOREST ALGO for ANOMALY/OUTLIER DETECTION here.

As this P.S. is entirely depenedent on OUTLIERS and OUTLIERS play an important role in this P.S.-: So; we use ISOLATION FOREST algo here for OUTLIER/ANOMALY DETETION.

```python
import matplotlib.pyplot as plt
plt.scatter(x=df.loc[:,'x'],y=df.loc[:,'y'])
```

SO; the 2 clusters are representation of records with NORMAL POINTS i.e. with NO DISEASE

And the RECORDS which are scattered entirely and ARE not part of any CLUSTER can be considered as OUTLIERS .i.e records having DISEASE.

It is not necessary that all scatterd data-points will have disease.
SO; to make sure that which scattered data-points have disease for that we will be using ISOLATION-FOREST for ANOMALY DETECTION.

```python
from sklearn.ensemble import IsolationForest
```

contamination parameter -:
default-value = "auto"




CONTAMINATION == THRESHOLD FOR ANOMALY SCORES of ALL DATA-POINTS IN THE DATASET.

if NO value for it is given then then Isoaltion FOrest based on anomaly socres of all data-points will select the "contamination" parameter value .i.e. THRESHOLD VALUE.

But; if we give the "contamination" parameter value then it hsould be between 0 (excluded) and 0.5 (included) always

```python
clf = IsolationForest(contamination=0.2)
```

If data-point has anomaly-score less than 0.2 => NORMAL POINT

If data-point has anomaly-score more than 0.2 => ANOMALY / OUTLIER POINT

contamination=0.2

```python
clf
```

```python
clf.fit(df)
```

```python
clf.predict(df)
```

Records with labels = "1" are  NORMAL-POINTS according to the Isolation Forest algo..

Records with labels = "-1" are  OUTLIER / ANOMALY-POINTS according to the Isolation Forest algo..

```python
predictions = clf.predict(df)
```

```python
import numpy as np
np.where(predictions < 0)
# Indices/Index of all data points that are classified as outliers .i.e. have predication label = -1 
# .i.e. less than 0
```

```python
abn_ind = np.where(predictions < 0)
```
