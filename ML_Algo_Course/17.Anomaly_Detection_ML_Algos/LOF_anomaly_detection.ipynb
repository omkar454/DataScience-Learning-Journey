# Local Outlier Factory (LOF) Anomaly Detection .M.L. algo

The anomaly score of each sample is called the Local Outlier Factor. It measures the local deviation of the density of a given sample with respect to its neighbors. It is local in that the anomaly score depends on how isolated the object is with respect to the surrounding neighborhood. More precisely, locality is given by k-nearest neighbors, whose distance is used to estimate the local density. By comparing the local density of a sample to the local densities of its neighbors, one can identify samples that have a substantially lower density than their neighbors. These are considered outliers.

# Widely used Parameters and Attributes of LOF .M.L. Algo-:





'kd_tree' does grid based searching .i.e. searches nearest grids w.r.t. the point instead of all the other points for calculation of k_neighbors for a particular POINT






default value of metric="minowski" and default value of p = 2. 

So; default metric distance indirectly = "euclidean" as "minowski" at p=2 is Euclidean distance metric only.



"contamination" parameter value is used by LOF algo. for setting threshold value for LOF score.

# ATTRIBUTES-:



when contamination param. value ='auto' then offset / threshold value for LOF score is "-1.5" by LOF algo. ALWAYS.



LOF detects outliers by comparing local density using distance (Minkowski or euclidean or manhattan), finds neighbors using efficient algorithms (KDTree/BallTree), and classifies points using a threshold (~ -1.5 in sklearn).

In sklearn, LOF uses negative scoring (negative_outlier_factor_), so more negative values indicate outliers, and thresholding is done using a negative offset (≈ -1.5) when contamination parameter = "auto" and NOT  a positive OFFSET.

threshold or offset value by sklearn for LOF score is always NEGATIVE and also LOF score for all data-points by LOF algo in sk-learn is always negative or 0.

```python
from sklearn.neighbors import LocalOutlierFactor
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
```

```python
X,y = make_circles(n_samples=750,factor=0.3,noise=0.1)

# We have introduced noise="0.1" as there should be some OUTLIERs/NOISE which will be DETECTED by LOF algo. while it does ANOMALY DETECTION.
```

```python
lof = LocalOutlierFactor(n_neighbors=5)
```

```python
plt.scatter(X[:,0],X[:,1],c=y)
```

```python
lof.fit_predict(X)
```

According to LOF Anomaly/Outlier detection Algo. -:

All the points which have got label as "-1" are OUTLIERS.

All the points which have got label as "1" are NOT OUTLIERS.

```python
lof.algorithm
```

```python
lof.contamination
```

```python
lof.offset_
```

```python
lof.negative_outlier_factor_

# this attribute gives the actual LOF score for each and every data-point in "NEGATIVE" value format.

# we did contamination="auto" -: So; threshold value by LOF = "-1.5" for this contamination property value = "auto". So. -:
# LOF-score ≥ threshold → Normal (1)
# LOF-score < threshold → Outlier (-1)
```

```python
lof.p
```

```python
plt.scatter(X[:,0],X[:,1],c=lof.fit_predict(X))
```

So; all the purple points are ANOMALIES/OUTLIERS according to LOF anomaly detection Unsupervised M.L. algorithm.







# 👉 sklearn flips LOF to negative so that “lower score = more anomaly”, and points below threshold (~ -1.5) are classified as outliers.
# This is done to make LOF anomaly detection algo consistent with other anomaly detection algo's like Isolation-Forest.
