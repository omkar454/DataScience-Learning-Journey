DBSCAN Clustering can do NON-LINEAR CLUSTERING ALSO.

So; this CLUSTERING ALGO. can be used to do ANOMALY/OUTLIER-DETECTION also as it captures and clusters the ENTIRE DATASET in NON-LINEAR SEPERABLE MANNER also.

So; in those P.S. where the OUTLIER DETECTION is MUST & OUTLIERS play a VERY IMPORTANT ROLE in those P.S. and DATASET we can apply DB-SCAN clustering algo. for ANOMALY/OUTLIER DETECTION from the given dataset.

So; DBSCAN can create CLUSTERS from NON-LINEAR DATA due to this property of DB-SCAN -: It can be used for OUTLIER/ANOMALY DETECTION also

```python
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
```

Widely used params of make_circles-:


✅ In short:

factor → controls relative size of inner circle

noise → controls amount of randomness/distortion

```python
X,y = make_circles(n_samples=750,factor=0.3,noise=0.1)

# We have introduced noise="0.1" as there should be some OUTLIERs/NOISE which will be DETECTED by DB_SCAN algo. while it does ANOMALY DETECTION.
```

```python
X
```

```python
y
```

```python
plt.scatter(x=X[:,0],y=X[:,1],c=y)
```

```python
from sklearn.cluster import DBSCAN
```

```python
dbscan = DBSCAN(eps=0.10)
```

```python
dbscan.fit_predict(X)
```

All the data-points with label as "-1" are NOISE-POINTS .i.e. are PREDICTED as ANOMALY / OUTLIER by DB-SCAN while doing ANOMALY DETECTION.

So; DB-SCAN is able to distinguish the points (NOISE-POINTS / OUTLIERS) which are COMPLETELY DIFFERENT from the NORMAL POINTS (CORE & BORDER POINTS).

So; It is able to predict/detect-: ANOMALY/OUTLIERS.

```python
dbscan.labels_
```

```python
plt.scatter(X[:,0],X[:,1],c=dbscan.labels_)
```

So; the purple points on this graph are classifed as OUTLIERS by DBSCAN.

So; DB-SCAN due to its NON-LINEAR CLUSTERING property is able to capture and detect all the ABNORMAL DATA-POINTS  and able to distinguish these ABNORMAL DATA-POINTS(OUTLIERS) from NORMAL DATA-POINTS SUCCESSFULLY.

```python
plt.scatter(x=X[:,0],y=X[:,1],c=y)
```

So; DBSCAN is an effective algorithm for anomaly/outlier detection. It identifies outliers as points that do not belong to any dense cluster. The eps (epsilon) hyperparameter plays a crucial role, as it defines the neighborhood radius. By adjusting eps, we can control how clusters are formed and consequently which points are detected as outliers, making the results dependent on the dataset and the chosen value for "eps" parameter and also depending on the P.S. and its situation.
