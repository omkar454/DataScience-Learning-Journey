```python
# For target feature/variable [O/P Feature]-:

# 👉 Higher number = more disease progression
# 👉 Lower number = less disease progression

# It does NOT mean “more diabetic” directly,
# but it means diabetes condition has worsened more after one year.

# In short:
# Higher value → disease progressed more.
```

```python
import pandas as pd
```

```python
df_diabetes = pd.DataFrame(data=dataset['data'],columns=dataset['feature_names'])
```

```python
df_diabetes.head(5)
```

```python
# Independent and Dependent features-: 
X = df_diabetes
y = dataset['target']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=10)
```

```python
X_train.head()
```

```python
# correlation-:
X_train.corr()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(15,10))
sns.heatmap(X_train.corr(),annot=True)
plt.show()
```

```python
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor()
regressor.fit(X_train,y_train)
```

```python
y_pred = regressor.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
# ORDER--> y_test then y_pred in terms of positional arguments positioning
print(r2_score(y_pred=y_pred,y_true=y_test))
print(mean_absolute_error(y_pred=y_pred,y_true=y_test))
print(mean_squared_error(y_pred=y_pred,y_true=y_test))
```





👉 Negative R² = biggest red flag here.

```python
# By default parameters of DT regressor this will easily lead to overfitting and performance metrics is also very BAD here.
# So; we will do HYPER-PARAMETER TUNING of Decision Tree Regressor .i.e will Perform PRE-PRUNING PROCESS here-:
```

# HyperParameter Tuning

```python
param = {
    # varaince reduction ===  'friedman_mse'
    'criterion':['squared_error','friedman_mse','absolute_error'],
    'splitter':['best','random'],
    'max_depth':[1,2,3,4,5,10,15,20,25],
    'max_features':['sqrt','log2']
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
regressor = DecisionTreeRegressor()
```

```python
grid = GridSearchCV(estimator=regressor,param_grid=param,cv=5,scoring='neg_mean_squared_error',verbose=3)
# 'neg_mean_squared_error' == Variance Reduction
```

```python
grid.fit(X_train,y_train)
```

Reason of few failed cases/fits-:




```python
grid.best_params_
```

```python
grid
```

```python
grid.best_estimator_
```

```python
# predictions-:
y_pred = grid.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
# ORDER--> y_test then y_pred in terms of positional arguments positioning
print(r2_score(y_pred=y_pred,y_true=y_test))
print(mean_absolute_error(y_pred=y_pred,y_true=y_test))
print(mean_squared_error(y_pred=y_pred,y_true=y_test))
```

```python
from sklearn import tree
import matplotlib.pyplot as plt
plt.figure(figsize=(25,15))
print(tree.plot_tree(grid.best_estimator_,filled=True))
```

Due to HyperParameter Tuning of DT-Regressor .i.e. Pre-pruning process on DT we got a better positive R square and lower MAE and MSE as compared to the DEFAULT Parameters of DT-Regressor .i.e. without PRE-PRUNING.
