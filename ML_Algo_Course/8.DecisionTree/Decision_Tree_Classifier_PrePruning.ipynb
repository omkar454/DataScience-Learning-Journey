```python
# train test split-:
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=10)
```



👉 Rule:

If algorithm uses distance or gradient → scaling needed

If algorithm uses tree-based splits → scaling not needed

```python
# apply Decison Tree classifier-: 
from sklearn.tree import DecisionTreeClassifier
treeclassifer = DecisionTreeClassifier()
```

```python
treeclassifer.fit(X_train,y_train)
```

```python
# Visualize the Decision Tree-: 
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(treeclassifer,filled=True)
plt.show()
#  x[0] => sepal length => DEPENDENT FEATURE at 0th index of X
#  x[1] => sepal width => DEPENDENT FEATURE at 1th index of X
# x[2] => petal length => DEPENDENT FEATURE at 2th index of X
# x[3] => petal width => DEPENDENT FEATURE at 3th index of X
```

Creating this type of Decision Tree with default parameters will always lead to OVERFITTING as the D.T. will be created till very DEPTH for the entire training data.

```python
y_pred = treeclassifer.predict(X_test)
from sklearn.metrics import confusion_matrix,classification_report
cm = confusion_matrix(y_pred,y_test)
print(cm)
print(classification_report(y_pred,y_test))
```

This would lead to OVERFITTING. So; to avoid it we can do -:
1) Post Pruning
2) Pre Pruning

# 1) Post Pruning

By looking at diagram we observed that after depth = 2 => we do NOT need to do more splitting cause split becomes almost PURE at depth = 2 on both left and right child.

So; we will do POST-PRUNING by setting max_depth = 2

```python
treeclassifer = DecisionTreeClassifier(max_depth=2)
treeclassifer.fit(X_train,y_train)
plt.figure(figsize=(15,10))
tree.plot_tree(treeclassifer,filled=True)
plt.show()
```

This is a POST-PRUNING TECHNIQUE

```python
y_pred = treeclassifer.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
cm = confusion_matrix(y_pred,y_test)
print(cm)
print(classification_report(y_pred,y_test))
```

With post-pruning-: 2 points were missclassified w.r.t test data
without post-pruning-: 1 point was missclassifed w.r.t test data

But; without post-pruning we did all the splits till highest depth still we got 1 wrongly predicted point-: there is nothing better we could do than this for creation of .D.T. still 1 point missclassifed w.r.t. test-data.

With post-pruning -: only 2 missclassified points we got w.r.t test-data

So; post-pruning is better instead of creating a D.T. with default parameters .i.e. without post-pruning leading to OVER-FITTING.

## Decision Tree Classifier Prepruning And Hyperparameter Tuning

```python
# Always give the value in the form of LIST consisting of different values for that particular parameter in order to test it via GridSearchCV/RandomSearchCV and also key names SHOULD BE EXACTLY same as  the parameters name and spelling which is in documentation of that particular MODEL.
param_grid = {
    "criterion":['gini','entropy','log_loss'],
    "splitter":["best","random"],
    "max_depth":[1,2,3,4,5],
    "max_features":["sqrt","log2"]
}

# NULL in JAVA == None in Python
# np.nan == Not a Number
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
treemodel = DecisionTreeClassifier()
```

```python
grid = GridSearchCV(estimator=treemodel,param_grid=param_grid,cv=5,scoring='accuracy')
```

```python
# This below code of warnings module will remove the warnings from             the O/P logger.
import warnings
warnings.filterwarnings('ignore')
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
# got 95% Accuracy for pre-pruning technique that is by doing HYPERPARAMETER-TUNING on "Decision Tree Classifier" model
```

```python
y_pred = grid.predict(X_test)
```

```python
y_pred
```

```python
cm = confusion_matrix(y_pred,y_test)
print(cm)
print(classification_report(y_pred,y_test))
```

```python
from sklearn.metrics import accuracy_score
score=accuracy_score(y_test,y_pred)
```
