```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
```



Crescent means a curved shape like a half moon 🌙.

```python
X,y = make_moons(n_samples=250,noise=0.05)
```




make_moons() generates exactly 2 clusters. The noise parameter does not represent percentage; it controls how much randomness is added to the data points.

```python
plt.scatter(x=X[:,0],y=X[:,1])
```

```python
plt.scatter(x=X[:,0],y=X[:,1],c=y)
```

Now we will do cluster these datapoints which are representation of NON-LINEAR RELATIONSHIPS / DATA-POINTS by using DBSCAN-Clustering Algo. and using its one PARAMETER -: "eps"


```python
# Since; DBSCAN is a distance based clustering algo.. So; need to bring the entire DATASET into same range/scale by doing STANDARDIZATION-:
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
# Standardization only happens to I/P features only and NOT to O/P feature as we want O/P prediciton always in its original scale/range only.
X_scaled = scaler.fit_transform(X)
```



After standardizing the dataset, eps becomes easier to choose because all features are on the same scale, making distance calculations more reliable.

```python
dbscan = DBSCAN(eps=0.3)
```

Widely used params of DBSCAN-:



```python
dbscan.fit(X_scaled)
```

Widely used attributes after doing CLUSTERING done by DBSCAN -:



```python
dbscan.labels_
```

"-1" label means that datapoint is an OUTLIER / NOISE-POINT
.So;  It has NOT been assigned to any CLUSTER by DBSCAN.
This is how efficiently DBSCAN handles OUTLIERS

```python
plt.scatter(X[:,0],X[:,1],c=dbscan.labels_)
```

The black point on graph is an OUTLIER. It has NOT been assigned to any CLUSTER by DB-SCAN.

```python
plt.scatter(X[:,0],X[:,1],c=y)
```

In real world scenario-: you will never get "y" O/P feature which will represent the cluster-labels for DATA-POINTS. You will only get X .i.e. I/P features and based on these features do CLUSTERING and find HIDDEN PATTERNS and BASED on "SILHOUTEE SCORING" -: Performance metrics for Clustering -: You will get optimal PARAMETERS for theat Clustering Algo.. and then get OPTIMAL CLUSTERING RESULTS after doing HYPER-PARAMETER TUNING and checking each results with "SILHOUTEE SCORING" which is a Performance metrics for Clustering 

But since this was a SYNTHETIC DATASET of make_moon method and it gave the feature representing cluster-labels for DATA-POINTS -: So; we can use it for VALIDATION of PERFORMANCE by DBSCAN-Clusteing on "X" I/P features DATASET.

So; DBSCAN gave almost same CORRECT CLUSTERING w.r.t. "y" O/P of cluster lable feature with ONLY ONE POINT categorized as OUTLIER by DBSCAN algo..

 REMAINIG CLUSTERING OF DBSCAN IS SAME w.r.t. "y" O/P feature.
