### Assignment 13: Sum of Digits

Write a program that calculates the sum of the digits of a number input by the user using a while loop.

```python
num = int(input("Enter a number:"))
number = num
sum_dig = 0
while num > 0:
    digit = num % 10
    sum_dig += digit
    num = num // 10
print(f"Sum of digits of {number} is: {sum_dig}")
```
