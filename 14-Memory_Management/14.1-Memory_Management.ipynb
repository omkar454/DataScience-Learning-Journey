````python
# Get unreachable objects-:
gc.set_debug(gc.DEBUG_SAVEALL)
# gc.set_debug(gc.DEBUG_SAVEALL) means“Do NOT delete unreachable objects — save them in gc.garbage so I can inspect them.”

gc.collect()
print(len(gc.garbage))   # shows unreachable objects
````

````python
import gc

# ✔ __del__ runs only when reference count becomes 0

class MyObject:
    def __init__(self,name):
        self.name = name
        print(f"Object {self.name} created")

    def __del__(self):
        print(f"Object {self.name} deleted")

# Create circular reference-:
obj1 = MyObject("obj1")
obj2 = MyObject("obj2")

obj1.ref = obj2
obj2.ref = obj1

del obj1
del obj2

# Python automatically deletes objects when their reference count becomes 0.
# Objects involved in circular references may not be deleted automatically, especially if they have a __del__ method.
# Calling gc.collect() forces the garbage collector to safely detect and delete unreachable objects, running __del__ where possible.

## Manually trigger the garbage collection
gc.set_debug(gc.DEBUG_SAVEALL)
gc.collect()
````

````python
# gc.garbage shows unreachable objects detected in the latest garbage collection.
# It does not accumulate old objects from previous collections unless they were still unreachable.

print(len(gc.garbage))
````

````python
def generate_numbers(n):
    for i in range(n):
        yield i

# Using the generator-:
for num in generate_numbers(100000):
    print(num)
    if num>10:
        break
````

````python
# `tracemalloc` is a **built-in Python module** used to **trace memory allocations**.

# ### ✔ Simple meaning

# It helps you **see where your program is using memory**, which is very useful for:

# * Finding **memory leaks**
# * Debugging **high memory usage**
# * Tracking memory allocations **over time**

# ### ✔ How it works

# * You start tracing with `tracemalloc.start()`
# * Run your code
# * Take a **snapshot** to see memory usage per line or object
# * Compare snapshots to see **where memory increased**

# ### ✔ Example

# ```python
# import tracemalloc

# tracemalloc.start()

# # some memory-heavy code
# a = [i for i in range(100000)]

# snapshot = tracemalloc.take_snapshot()
# top_stats = snapshot.statistics('lineno')

# for stat in top_stats[:5]:
#     print(stat)
# ```

# This shows **top 5 lines consuming memory**.

# ### ✔ Short summary

# **`tracemalloc` is used to trace and debug memory allocations, helping detect memory leaks and high memory usage in Python programs.**
````

````python
# Profiling memory Usgae with tracemalloc-:

# Using tracemalloc with 'lineno' shows how much memory was allocated by each line of code individually.
# Multiple allocations on the same line are summed together.
# Different lines are reported separately, so you can see which lines use the most memory.

import tracemalloc

def create_list():
    return [i for i in range(10000)]

def main():
    tracemalloc.start()

    create_list()
    snapshot = tracemalloc.take_snapshot()
    top_stats = snapshot.statistics('lineno')

    print("[Top 10]")
    for stat in top_stats[:10]:
        print(stat)
````

````python
main()
````
