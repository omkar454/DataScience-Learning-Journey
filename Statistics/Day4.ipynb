### Standardization

```python
import pandas as pd
import seaborn as sns
```

```python
df = sns.load_dataset('titanic')
df.head()
```

```python
df.isnull().sum()
```

```python
df['age'].fillna(df.age.median(),inplace=True)
```

```python
df.isna().sum()
```

```python
df = df[['survived','pclass','age','fare']]
# Independent and Dependent Features-:
df
```

```python
# Independent and Dependent Features-:
X = df.iloc[:,1:] #Independent
Y = df.iloc[:,0]  #dependent
```

```python
X
```

```python
# train_test_split is used to split a dataset into training and testing sets.
# Training set → used to train the model
# Test set → used to evaluate model performance
# 📌 Commonly used in machine learning to check how well a model generalizes to new data.

# from sklearn.model_selection import train_test_split
# X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
# 👉 Here, 80% data for training and 20% for testing.
```

```python
# It is very important to do train_test_split cuase we don't want to overfit the Model.

# Overfitting means the model memorizes the training data instead of learning patterns, so it works well on training data but fails on new data.
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.33,)
```
