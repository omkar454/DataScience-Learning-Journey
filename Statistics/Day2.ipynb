```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
dataset = [11,10,12,14,12,15,14,13,15,102,12,14,17,19,107,10,13,12,14,12,108,12,11,14,13,15,10,15,12,19,14,13,15,10]
```

```python
plt.hist(dataset)
```

```python
sns.histplot(dataset,kde=True)
```

```python
mean=np.mean(dataset)
std = np.std(dataset)
z_distr=[]
for i in dataset:
        z_score=(i-mean)/std
        z_distr.append(z_score)
print(z_distr)
```

```python
# Z Score
outliers=[]
def detect_outliers(data):
    threshold=3
    mean=np.mean(data)
    std = np.std(data)

    for i in data:
        z_score=(i-mean)/std
        if np.abs(z_score) > threshold:
            outliers.append(i)
    return outliers
```

```python
detect_outliers(dataset)
```

```python
# IQR-:(InterQuartile Range)
'''
1. Sort the data
2. Calculate Q1 and Q3
3. IQR (Q3 - Q1)
4. Find the Lower fence (Q1-1.5(IQR))
5. Find the Upper fence (Q3+1.5(IQR))
'''
# NumPy automatically sorts the data internally when you use np.percentile().
dataset = sorted(dataset)
q1,q3 = np.percentile(dataset,[25,75])
iqr = q3 - q1
LowerFence = q1 - (1.5*iqr)
HigherFence = q3 + (1.5*iqr)
```

```python
print(LowerFence, HigherFence)
```

```python
sns.boxplot(dataset)
```

```python
for i in dataset:
    if i < float(LowerFence ) or i > float(HigherFence):
       dataset.remove(i)
print(dataset)
```
