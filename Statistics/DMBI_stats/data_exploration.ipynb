```python
import numpy as np
import pandas as pd
import seaborn as sns
```

# Q1

```python
ds = np.array([57 , 57, 57,58,63,66,66,67,67,68,69,70,70,70,72,73,75,75,76,76,78,79,81 ])
ds.sort()
```

```python
print(ds)
```

```python
print(f"mean:{np.mean(ds)}")
print(f"median:{np.median(ds)}")
```

```python
vals, counts = np.unique(ds,return_counts=True)
```

```python
vals,counts
```

```python
counts == counts.max()
```

```python
print(f"Mode: {vals[counts == counts.max()]}")
```

```python
maximum = ds.max()
minimum = ds.min()
maximum,minimum
```

```python
midrange = (maximum+minimum)/2
print(f"midrange:{midrange}")
range_ds = maximum - minimum
print(f"range:{range_ds}")
```

```python
zero_percentile = np.quantile(ds,0)
print(f"0th pecentile: {zero_percentile}")
Q1 = np.quantile(ds,0.25)
print(f"25th pecentile: {Q1}")
median = np.quantile(ds,0.50)
print(f"50th pecentile: {median}")
Q3 = np.quantile(ds,0.75)
print(f"75th pecentile: {Q3}")
maximum = np.quantile(ds,1)
print(f"100th pecentile: {maximum}")
```

```python
IQR = Q3 - Q1
print(f"InterQuartile Range: {IQR}")
```

```python
lower_fence = Q1 - 1.5*IQR
higher_fence = Q3 + 1.5*IQR
```

```python
print(f"Lower fence:{lower_fence}\nHigher fence:{higher_fence}")
```

BOXPLOT

```python
sns.boxplot(ds)
```

```python
ds
```

Since all the points in dataset are greater than lower fence value and lower than higher fence value.
So; there are NO OUTLIERS in datatset
