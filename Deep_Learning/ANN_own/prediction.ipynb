```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

# Load ANN trained H5 model, scaler & OHE pickle model 

```python
model = load_model('model.h5')
with open('onehot_encoder_geo.pkl','rb') as file:
    onehot_encoder_geo = pickle.load(file)
    
with open('label_encoder_gender.pkl','rb') as file:
    label_encoder_gender = pickle.load(file)

with open('scaler.pkl','rb') as file:
    scaler = pickle.load(file)
```

```python
# Example input data
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
input_data['Gender']=label_encoder_gender.transform([input_data['Gender']])[0]
```

```python
for i in onehot_encoder_geo.get_feature_names_out():
    print(i)
```

```python
onehot_encoder_geo.transform([[input_data['Geography']]]).toarray()
```

```python
k=0
for i, j in zip(
    onehot_encoder_geo.get_feature_names_out(),
    onehot_encoder_geo.transform([[input_data['Geography']]]).toarray().flatten()
):
    input_data[i] = j
```

zip() pairs the feature names with their corresponding encoded values, so each name gets its correct 0 or 1.

Example: ['France','Germany'] + [1,0] → ('France',1), ('Germany',0).
