```python
#Syntax
lambda arguments: expression
```

```python
add = lambda a,b:a+b
print(type(add))
print(add(2,3))
```

```python
even = lambda a:a%2==0
print(even(21))
```

```python
add3 = lambda a,b,c:a+b+c
print(add3(12,13,14))
```

```python
numbers = [1,2,3,4,5,6]
list(map(lambda x:x**2,numbers))
```

```python
print(map(lambda x:x**2,numbers))
for i in map(lambda x:x**2,numbers):
    print(i)
```
