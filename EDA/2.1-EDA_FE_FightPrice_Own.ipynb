````python
# Feature Engineering-:
# Pandas doesn’t apply string functions directly on columns, so .str is used to tell pandas “this is text — apply this string operation to every value.”
df['Date'] = df["Date_of_Journey"].str.split("/").str[0]

# .str converts the column into a string-operations mode, split("/") breaks each value, and .str[0] selects the first split part from every row, storing it in Date.

# .str[0] means “take the first element from each value in the column.”
# So after splitting, it picks the first part of every row’s list (not the first row).
````

````python
# alone [0] would try to get the first row of the column,
# but .str[0] gets the first part of every split value in each row.

# So:
# alone [0] → first row
# .str[0] → first element of each split string in all rows
````

````python
# ## ⚠️ SettingWithCopyWarning in pandas

# ### **What it means**

# This warning occurs when you try to modify a **slice of a DataFrame**, which might be a **copy**, not the original DataFrame.
# Your change may not affect the actual DataFrame.

# ### **Problematic example**

# ```python
# df.iloc[6474]["Duration"] = '0h 5m'  # ❌ Warning
# ```

# ### **Recommended method**

# Use **`.loc`** for direct assignment:

# ```python
# df.loc[6474, "Duration"] = '0h 5m'  # ✅ Safe
# ```

# **Key point:**

# * `.loc[row_index, column_name]` updates the original DataFrame directly and avoids the warning.
````
