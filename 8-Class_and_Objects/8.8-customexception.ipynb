# Custom exception (Raise and Throw an Exception)

```python
class Error(Exception):
    pass

class dobException(Error):
    pass
```

```python
year = int(input("Enter the dob:"))
age = 2025 - year
try:
    if age<=20 and age >=30:
        print("The age is valid so you can apply for exams")
    else:
        raise dobException
except dobException:
    print("Sorry your age is NOT within the age limit")
```

```python
age = -5
if age < 0:
    raise ValueError("Age cannot be negative")
```

```python
def divide(a, b):
    if b == 0:
        raise ZeroDivisionError("Cannot divide by zero")
    return a / b
```

```python
divide(20,0)
```

```python
class InvalidInputError(Exception):
    pass

raise InvalidInputError("This is a custom error!")
```
