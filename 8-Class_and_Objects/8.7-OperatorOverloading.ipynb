Operator Overloading

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self,other):
        return Vector(self.x+other.x, self.y+other.y)

    def __sub__(self,other):
        return Vector(self.x-other.x, self.y-other.y)

    def __mul__(self,other):
        return Vector(self.x*other,self.y*other)

    def __eq__(self,other):
        return self.x == other.x and self.y == other.y

    def __repr__(self):
        return f"Vector({self.x},{self.y})"

v1 = Vector(2,3)
v2 = Vector(4,5)

print(v1+v2)
print(v1*3)
print(v1-v2)
print(v1==v2)
print(repr(v1))
```

```python
class ComplexNumber:
    def __init__(self, x,y):
        self.x = x
        self.y = y

    def __add__(self,other):
        return ComplexNumber(self.x+other.x, self.y + other.y)
    
    def __sub__(self,other):
        return ComplexNumber(self.x-other.x, self.y - other.y)

    def __mul__(self,other):
        return ComplexNumber(((self.x*other.x )- (self.y*other.y)),
        ( (self.y * other.x) + (self.x*other.y)))

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __repr__(self):
        return f"{self.x}+({self.y})i"
    
    def __truediv__(self,other):
        denominator = ((other.x)**2 + (other.y)**2)
        true_part = ((self.x*other.x) + (self.y*other.y))/denominator
        imag_part =( (self.y*other.x) - (self.x*other.y))/denominator
        return ComplexNumber(true_part,imag_part)

    
c1 = ComplexNumber(2, 3)
c2 = ComplexNumber(1, 4)

print(c1+c2)
print(c1-c2)
print(c1*c2)
print(c1/c2)
print(c1==c2)
print(c2)
```
