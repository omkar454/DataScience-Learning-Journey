```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def __str__(self):
        return f"{self.name},{self.age} years old"
    
    def __repr__(self):
        return f"Person(name={self.name}, age={self.age})"

    def __call__(self):
        print(f"{self} was called")

print((Person("Omkar",20)))
print(repr(Person("Omkar",20)))
```

```python
Person("Omkar",20)()
```

```python
class Team:
    def __init__(self, players):
        self.players = players
    
    def __len__(self):
        return len(self.players)

t = Team(["a","b","c"])
print(len(t))   # 3
```

```python
class Box:
    def __init__(self, items):
        self.items = items

    def __getitem__(self, index):
        return self.items[index]

b = Box([1, 2, 3])
print(b[1])   # 2
```

```python
class Box:
    def __init__(self, items):
        self.items = items

    def __setitem__(self, index, value):
        self.items[index] = value

b = Box([1,2,3])
b[1] = 10
print(b.items)   # [1, 10, 3]
```
