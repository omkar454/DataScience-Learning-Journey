```python
class Dog:
    # constructor
    def __init__(self,name,age):
        self.name = name
        self.age= age

dog1 = Dog("Oreo",3)
print(dog1)
print(dog1.name)
print(dog1.age)
```

```python
dog2 = Dog("Lucy",4)
print(dog2.name)
print(dog2.age)
```

```python
class Dog:
    def __init__(self, name, age):
        # Instance variables
        self.name = name
        self.age = age

    # Instance methods-:
    def bark(self):
        print(f"{self.name} is Barking. WOOF!!!!")

dog1 = Dog("Buddy", 3)
dog1.bark()
```

```python
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"₹{amount} is Deposited. New balance is: ₹{self.balance}")
        
    def withdraw(self,amount):
        if amount > self.balance:
            print(f"Insufficient Funds. Available Amount: ₹{self.balance}")
        else:
            self.balance -= amount
            print(f"Amount Wtihdrawn successfully.Available Amount: ₹{self.balance}")
    
    def get_balance(self):
        return f"₹{self.balance}"
```

```python
account = BankAccount("Omkar",10)
print(account.get_balance())
account.deposit(1000)
```

```python
print(account.get_balance())
account.withdraw(10000)
```
