```python
class Person:
    def __init__(self, name, age, gender):
        #Protected variabels-: _
        self._name = name 
        self._age = age
        self.gender = gender

class Employee(Person):
    def __init__(self, name, age, gender):
            super().__init__(name,age,gender)

employee = Employee("Krish", 34, "Male")
print(employee._name)
```

```python
print(Person("Omkar",20,"M")._name)
```

```python
# Encapsulation with Getter and Setter-:
class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age
    # Getter-:
    def get_name(self):
        return self.__name
    # Setter-:
    def set_name(self, name):
        self.__name = name
        
    def get_age(self):
        return self.__age
    
    def set_age(self, age):
        if age > 0:
         self.__age = age
        else:
            print("Age cannot be negative.")


person = Person("Omkar", 20)
print(person.get_name())
person.set_name("Raut")
print(person.get_name())

print(person.get_age())
person.set_age(4)
print(person.get_age())
person.set_age(-14)


```
