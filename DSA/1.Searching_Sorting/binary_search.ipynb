```python
def binary_search(arr,target):
    start = 0; end = len(arr) - 1
    for  i in range(len(arr)):
        while end>=start:
            mid = (end+start)//2
            if arr[mid] == target:
                return mid
            elif arr[mid] > target:
                end = mid-1
            elif arr[mid] < target:
                start = mid + 1
        return -1
```

```python
binary_search([1,2,3],23)
```
