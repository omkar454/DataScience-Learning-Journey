**Question 18:** Write a Python program to reverse a string.

```python
# Your code here
len = 0
reverse_str=""
str1 = "Omkar"
for i in str1:
    len += 1


for i in range(len-1,-1,-1):
    reverse_str += str1[i]
print(reverse_str)
```

**Question 19:** Write a Python program to check if a string is a palindrome.

```python
# Your code here
len = 0
reverse_str=""
str1 = "Madam"
for i in str1:
    len += 1


for i in range(len-1,-1,-1):
    reverse_str += str1[i]


if (reverse_str.lower() == str1.lower()):
    print("It is pallindrome")
```
